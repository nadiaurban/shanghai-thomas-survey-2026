## First, choose the colors you want to use for this part!

```python
import pprint
import matplotlib.colors as mcolors

colors = sorted(mcolors.CSS4_COLORS.keys())
pprint.pprint(colors, width=100)
```

```python
colors = [
    'pink',
    'magenta',
    'coral',
    'deeppink', 
    'darkviolet'
]

def show_palette(colors, title="Palette"):
    """Display a horizontal series of colored bars to visualize a palette."""
    fig, ax = plt.subplots(figsize=(len(colors) * 1.5, 1.5))
    for i, color in enumerate(colors):
        ax.barh(0, width=1, left=i, color=color)
    ax.set_xlim(0, len(colors))
    ax.set_ylim(-0.5, 0.5)
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_title(title, fontsize=14)
    plt.show()

# Visualize the updated palettes
show_palette(colors)
```

### 1. Simple bar chart

```python

QX_percentages = survey['QX_text'].value_counts(normalize=True) * 100
QX_percentages.sort_index().plot(kind='bar', color=colors)
plt.xlabel('XXX')
plt.ylabel('XXX')
plt.title('XXX')
plt.show()
```

### 2. Simple pie chart

```python

QX_percentages = survey['QX_text'].value_counts(normalize=True) * 100
plt.pie(QX_percentages, labels=QX_percentages.index, colors=colors, autopct='%1.1f%%')
plt.title('XXX')
plt.show()

```

###  3.Grouped by grade


```python

QX_grade = (survey.groupby('Q1_text')['QX_text'].value_counts(normalize=True).unstack() * 100).round(0)
QX_grade.plot(kind='bar', color=colors)
plt.xlabel('XXX')
plt.ylabel('XXX')
plt.title('XXX')
plt.legend(title="XXX")
plt.show()
```

### 4. Grouped by gender 

```python
XX_gender = (survey.groupby('Q2_text')['XX_text'].value_counts(normalize=True).unstack() * 100).round(1)

XX_gender.plot(kind='bar', color=colors)
plt.xlabel('XXX')
plt.ylabel('XXX')
plt.title('XXXX')
plt.legend(title='XXX')
plt.show()
```
