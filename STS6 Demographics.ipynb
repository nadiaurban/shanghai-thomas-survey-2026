## First, choose the colors you want to use for this part!

```python
import pprint
import matplotlib.colors as mcolors

colors = sorted(mcolors.CSS4_COLORS.keys())
pprint.pprint(colors, width=100)
```

```python
colors = [
    'cyan',
    'darkblue',
    'darkcyan',
    'turquoise', 
    'skyblue'
]

def show_palette(colors, title="Palette"):
    """Display a horizontal series of colored bars to visualize a palette."""
    fig, ax = plt.subplots(figsize=(len(colors) * 1.5, 1.5))
    for i, color in enumerate(colors):
        ax.barh(0, width=1, left=i, color=color)
    ax.set_xlim(0, len(colors))
    ax.set_ylim(-0.5, 0.5)
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_title(title, fontsize=14)
    plt.show()

# Visualize the updated palettes
show_palette(colors)
```

# Part 1: Simple bar charts and pie charts. 

## Question 1: Grades

What is the number of students in each grade? 

```python
# Q1: Grade Distribution (Bar Chart)
Q1_counts = survey['Q1_text'].value_counts()
Q1_counts.sort_index().plot(kind='bar', color=colors)
plt.xlabel('Grade')
plt.ylabel('Number of students in each grade')
plt.title('Grade Distribution')
plt.show()
```

```python
#Think: Are there big differences between each grade?
```

## Question 2 (Gender)


```python
# Q2: Gender Distribution (Bar Chart)
Q2_percentages = survey['Q2_text'].value_counts(normalize=True) * 100
Q2_percentages.sort_index().plot(kind='bar', color=colors)
plt.xlabel('Gender')
plt.ylabel('Percentage')
plt.title('Gender distribution of students in Shanghai Thomas School')
plt.show()
```

```python
# Think: Are there big differences between genders in our school?
```

## Question 3: Living situation

```python
# Q3: Living Situation (Pie Chart)
QX_percentages = survey['QX_text'].value_counts(normalize=True) * 100
plt.pie(QX_percentages, 
        labels=QX_percentages.index, 
        colors=colors, 
        autopct='%1.1f%%')
plt.title('Living Situation')
plt.show()

```

```python
# Think: What's the living situation of majority of our students?
```

```python
# Q4: Do You Have Siblings? (Pie Chart)
QX_percentages = survey['QX_text'].value_counts(normalize=True) * 100
plt.pie(QX_percentages, 
        labels=QX_percentages.index, 
        colors=colors, 
        autopct='%1.1f%%')
plt.title('Do you have any sibilings?')
plt.show()
```

```python
# Think: What's the family situation of majority of our students?
```

# Part 2: Grouping data 

## In this part we will group the answers by grades (`Q1_text`) and gender (`Q2_text`)



```python
# Grouping Q3 (Living situation) by Q1 (Grade)
Q3_grade = (survey.groupby('Q1_text')['Q3_text'].value_counts(normalize=True).unstack() * 100).round(0)
Q3_grade.plot(kind='bar', color=colors)
plt.xlabel('Grade')
plt.ylabel('Percentage')
plt.title('Number of Siblings by Grade')
plt.legend(title="Number of Siblings")
plt.show()
```

```python
# Group Q3 (Living Situation) by Q2 (Gender)
XX_gender = (survey.groupby('Q2_text')['XX_text'].value_counts(normalize=True).unstack() * 100).round(1)

XX_gender.plot(kind='bar', color=colors)
plt.xlabel('Gender')
plt.ylabel('Percentage')
plt.title('XXXX')
plt.legend(title='XXX')
plt.show()
```
